```python
class Person:
    def _init_(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

# 2. Define child class 'Student'
class Student(Person):
    def _init_(self, name, age, student_id):
        # 3. Use super() to initialize inherited attributes
        super()._init_(name, age)
        self.student_id = student_id

    # 5. Override display_info() to include unique ID
    def display_info(self):
        super().display_info()
        print(f"Student ID: {self.student_id}")

    def enroll_course(self, course):
        print(f"{self.name} has enrolled in {course}.")

# 2. Define child class 'Lecturer'
class Lecturer(Person):
    def _init_(self, name, age, employee_id):
        # 3. Use super() to initialize inherited attributes
        super()._init_(name, age)
        self.employee_id = employee_id

    # 5. Override display_info() to include unique ID
    def display_info(self):
        super().display_info()
        print(f"Employee ID: {self.employee_id}")

    def teach_course(self, course):
        print(f"{self.name} is teaching {course}.")

# 4. Create objects and demonstrate functionality
if _name_ == "_main_":
    # Create and demonstrate Student
    student = Student("Kwame Mensah", 20, "STU12345")
    print("--- Student Details ---")
    student.display_info()
    student.enroll_course("Database Systems")

    print("\n--- Lecturer Details ---")
    # Create and demonstrate Lecturer
    lecturer = Lecturer("Dr. Araba Serwaa", 45, "EMP98765")
    lecturer.display_info()
    lecturer.teach_course("Object-Oriented Programming")
```
